```python
def get_3mer_dict(seq: str) -> dict:
    tmer_dict = {}
    
    for i in range(3, len(seq)):
        tmer = seq[i - 3:i]
        tmer_dict[tmer] = tmer_dict.get(tmer, 0) + 1

    return tmer_dict

def add_gene_to_ref(ref_dict: dict, gene_dict: dict, name: str) -> dict:
    for tmer in gene_dict.keys():
        ref_dict[tmer].setdefault(gene_dict[tmer], []).append(name)
    
    return ref_dict
```

```python
from itertools import product

def initialize_3mer_dict():
    bases = "ATCG"
    tmer_dict = {}

    for mer in product(bases, repeat=3):
        tmer = "".join(mer)
        tmer_dict[tmer] = {}

    return tmer_dict
```

```python
import random

def rand_seq() -> str:
    all_3mers_sequence = []
    for _ in range(500):
        all_3mers_sequence += list("".join(
            "".join(codon)
            for codon in product("ACGT", repeat=3)
        ))
    random.shuffle(all_3mers_sequence)
    return "".join(all_3mers_sequence)

print(rand_seq())
```

```python
from collections import Counter

def find_matches(ref_dict: dict, gene_dict: dict, tmer_bound: int, match_bound: int) -> list:
    match_list = []
    for tmer in gene_dict.keys():
        for num_tmer in range(max(0, gene_dict[tmer] - tmer_bound // 2), 
                              (gene_dict[tmer] + tmer_bound // 2) + 1):
            match = ref_dict[tmer].get(num_tmer)
            if match:
                match_list += match

    matches = []
    counts = Counter(match_list)
    print(counts)

    for gene, num_match in counts.items():
        if num_match in range(len(gene_dict.keys()) - match_bound, len(gene_dict.keys()) + 1):
            print(gene)
            matches.append(gene)
    
    return matches
```
